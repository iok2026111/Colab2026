```python
import pandas as pd
import numpy as np

inp1 = input("Enter x:")
inp2 = input("Enter y:")
x = int(inp1)
y = int(inp2)
Z1 = []
Z2 = []
Z3 = []
X = []
Y = []

if 0<=x<=10 and 0<=y<=10:
  z1 = 5*x + y
  z2 = x**2 + 5*y
  z3 = x**3 + y**2
  Z1.append(z1)
  Z2.append(z2)
  Z3.append(z3)
  X.append(x)
  Y.append(y)

df = pd.DataFrame({
    'X':X,
    'Y':Y,
    'Z1':Z1,
    'Z2':Z2,
    'Z3':Z3
})

df

```

```python
import pandas as pd

X = []
Y = []
Z1 = []
Z2 = []
Z3 = []

for x in range(11):
  for y in range(11):

    z1 = 5*x + y
    z2 = x**2 + 5*y
    z3 = x**3 + y**2
    Z1.append(z1)
    Z2.append(z2)
    Z3.append(z3)
    X.append(x)
    Y.append(y)

df = pd.DataFrame({
    'X':X,
    'Y':Y,
    'Z1':Z1,
    'Z2':Z2,
    'Z3':Z3
})
df





```

```python
import matplotlib.pyplot as plt

fig,(ax1, ax2, ax3) = plt.subplots(1,3, figsize = (12,4))
fig.suptitle('Set of Mathematical Functions')

ax1.scatter(df['X'], df['Z1'], s=10)
ax1.set_title('Function Z1 = 5*x + y')

ax2.scatter(df['X'], df['Z2'], s=10)
ax2.set_title('Function Z2 = x^2 + 5*y')

ax3.scatter(df['X'], df['Z3'], s=10)
ax3.set_title('Function Z3 = x^3 + y^2')

for ax in (ax1, ax2, ax3):
  ax.set_xlabel('X')
  ax.set_ylabel('Z')

plt.tight_layout()
plt.show()



```

```python
import numpy as np
import matplotlib.pyplot as plt



# Convert dataFrame into matrices:
X = df['X'].to_numpy().reshape(11, 11)
Y = df['Y'].to_numpy().reshape(11, 11)

Z1 = df['Z1'].to_numpy().reshape(11, 11)
Z2 = df['Z2'].to_numpy().reshape(11, 11)
Z3 = df['Z3'].to_numpy().reshape(11, 11)

# Create 3d plots
fig, (ax1, ax2, ax3) = plt.subplots(1,3, figsize=(12,4), subplot_kw={'projection': '3d'})

ax1.plot_surface(X, Y, Z1)
ax2.plot_surface(X, Y, Z2)
ax3.plot_surface(X, Y, Z3)

plt.tight_layout()
plt.show()

```
